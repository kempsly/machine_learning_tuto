## load_summarize_chain

Creates a summarization pipeline that connects the LLM, prompt, and documents.

## llm

The language model used to generate the summary.

## chain_type='stuff'

All documents are combined into a single prompt and sent to the LLM. This works well for small or medium documents.

## prompt

The instruction given to the model to guide how it should summarize the text.

## verbose=True

Displays internal steps of the chain execution for debugging.

## chain.run(docs)

Runs the summarization chain on the documents and returns the result.

## output_summary

Stores the final summary generated by the model.

## Flow

docs → prompt → llm → summary

### Map reduce to Summarize Large documents 

### Map reduce to Summarize Large documents

Map reduce is a summarization strategy used when documents are too large to fit into a single prompt.

It works in two steps:

## Map step

Each document chunk is summarized individually by the LLM.

## Reduce step

All intermediate summaries are combined and summarized again into a final answer.

## Code example

```python id="map1"
from langchain.chains.summarize import load_summarize_chain

chain = load_summarize_chain(
    llm,
    chain_type="map_reduce",
    verbose=True
)

output_summary = chain.run(docs)
output_summary
```

## Explanation

load_summarize_chain creates a summarization pipeline.

chain_type="map_reduce" splits the work into multiple steps:

* map step summarizes each chunk
* reduce step merges summaries into one final result

verbose=True shows intermediate steps during execution.

chain.run(docs) runs the full pipeline on the documents.

output_summary contains the final condensed summary.

## Flow

documents → split into chunks → map (summarize each chunk) → reduce (combine summaries) → final summary


````python
from langchain_classic.text_splitter import RecursiveCharacterTextSplitter
````

````python
docs
````

````python
final_documents=RecursiveCharacterTextSplitter(chunk_size=2000,chunk_overlap=100).split_documents(docs)
final_documents
````

````python
len(final_documents)
````

````python
chunks_prompt="""
Please summarize the below speech:
Speech:`{text}'
Summary:
"""
map_prompt_template=PromptTemplate(input_variables=['text'],
                                    template=chunks_prompt)
````

````python
final_prompt='''
Provide the final summary of the entire speech with these important points.
Add a Motivation Title,Start the precise summary with an introduction and provide the summary in number 
points for the speech.
Speech:{text}

'''
final_prompt_template=PromptTemplate(input_variables=['text'],template=final_prompt)
final_prompt_template
````

````python
summary_chain=load_summarize_chain(
    llm=llm,
    chain_type="map_reduce",
    map_prompt=map_prompt_template,
    combine_prompt=final_prompt_template,
    verbose=True
)

output=summary_chain.run(final_documents)
output
````

### Refine Chain For Summarization 

### Refine summarization

Refine is a summarization strategy used to progressively improve a summary by processing documents one by one.

It starts with an initial summary and then updates it step by step as it reads new chunks of text.

## How it works

Start with the first document chunk to create an initial summary
Then for each next chunk, the model refines the existing summary by adding new relevant information

## Code example

```python id="ref1"
from langchain.chains.summarize import load_summarize_chain

chain = load_summarize_chain(
    llm,
    chain_type="refine",
    verbose=True
)

output_summary = chain.run(docs)
output_summary
```

## Explanation

load_summarize_chain builds a summarization pipeline.

chain_type="refine" tells LangChain to:

* create an initial summary from the first chunk
* iteratively improve it using the next chunks

verbose=True shows intermediate refinement steps.

chain.run(docs) runs the process on all documents.

output_summary contains the final refined summary.

## Flow

first chunk → initial summary → update with next chunk → refine again → final improved summary


````python
chain=load_summarize_chain(
    llm=llm,
    chain_type="refine",
    verbose=True
)
output_summary=chain.run(final_documents)
output_summary
````

# 1. stuff

## When to use

Small or medium documents that fit inside the LLM context window

## How it works

All documents are combined into one prompt and sent to the model

## Pros

Fast
Simple
Good quality for small inputs

## Cons

Fails on large documents (token limit)

## Best for

Short PDFs, notes, articles

# 2. map_reduce

## When to use

Large documents or many files

## How it works

Each chunk is summarized separately (map) then combined (reduce)

## Pros

Scales to large data
Works with many documents
Parallelizable

## Cons

More expensive
Final summary can lose some detail

## Best for

Research papers, large PDFs, document collections

# 3. refine

## When to use

When you want higher quality summaries and can afford slower processing

## How it works

Starts with one summary, then improves it chunk by chunk

## Pros

More coherent final summary
Keeps continuity
Good detail retention

## Cons

Slower (sequential processing)
More API calls

## Best for

Legal documents, financial reports, high-quality summaries

# Simple decision guide

If small document → stuff
If large document → map_reduce
If quality matters most → refine


# Quick mental model

stuff = one shot summary
map_reduce = divide and summarize then merge
refine = build and improve step by step

# Real-world recommendation

In most production systems:

* default → map_reduce (best balance)
* high precision → refine
* quick demos → stuff
